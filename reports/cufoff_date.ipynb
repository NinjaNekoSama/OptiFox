```python
import pandas as pd
```

```python
# Read icustays data and generate new feature cutoff_date
icustays = pd.read_feather("../data/processed/icustays_20240602.feather")
icustays['cutoff_date'] = icustays['icu_intime'] + pd.Timedelta(hours=24)

# Apply the condition to adjust cutoff_date
icustays['cutoff_date'] = icustays.apply(
    lambda row: min(row['cutoff_date'], row['icu_outtime']),
    axis=1
)

# Generate the cutoff_los column by calculating the difference between cutoff_date and icu_intime
icustays['cutoff_los'] = icustays['cutoff_date'] - icustays['icu_intime']

# Convert cutoff_los to a decimal format representing the total number of days
icustays['cutoff_los'] = icustays['cutoff_los'].dt.total_seconds() / 86400

# Reordering the columns
cols = list(icustays.columns)

# Move cutoff_date and cutoff_los columns between icu_intime and icu_outtime
cols.insert(cols.index('icu_outtime'), cols.pop(cols.index('cutoff_date')))
cols.insert(cols.index('icu_outtime'), cols.pop(cols.index('cutoff_los')))
icustays = icustays[cols]

icustays
```
